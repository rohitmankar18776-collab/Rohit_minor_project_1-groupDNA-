```python

with open("hostel_bois.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print("Total Lines:", len(lines))
```

```python

messages = []

for line in lines:
    line = line.strip()

    if " - " not in line:
        continue

    try:
        timestamp, rest = line.split(" - ", 1)

        if ": " not in rest:
            continue

        sender, message = rest.split(": ", 1)

        messages.append({
            "timestamp": timestamp,
            "sender": sender,
            "message": message
        })

    except:
        continue

print("Parsed Messages:", len(messages))
print(messages[:5])
```

```python

participants = set()
message_count = {}

for msg in messages:
    sender = msg["sender"]

    participants.add(sender)
    message_count[sender] = message_count.get(sender, 0) + 1

print("=" * 50)
print("GROUP OVERVIEW")
print("=" * 50)

print("Total Messages :", len(messages))
print("Participants   :", len(participants))

print("\nMessages Per Person:\n")

for name, count in sorted(message_count.items(),
                          key=lambda x: x[1],
                          reverse=True):

    percentage = (count / len(messages)) * 100

    print(f"{name:<15}{count:>5} ({percentage:.2f}%)")
```

```python

day_count = {}
hour_count = {}

for msg in messages:

    timestamp = msg["timestamp"]

    date, time = timestamp.split(", ")

    hour = int(time.split(":")[0])

    day_count[date] = day_count.get(date, 0) + 1
    hour_count[hour] = hour_count.get(hour, 0) + 1

busiest_day = max(day_count, key=day_count.get)
busiest_hour = max(hour_count, key=hour_count.get)

print("\nBusiest Day :", busiest_day,
      "-", day_count[busiest_day], "messages")

print("Busiest Hour:",
      f"{busiest_hour:02d}:00 - {busiest_hour:02d}:59",
      "-", hour_count[busiest_hour], "messages")
```

```python

import numpy as np

# Sort participants
participants = sorted(list(participants))

# Assign each participant an index
person_index = {name: i for i, name in enumerate(participants)}

# Create heatmap (rows = participants, columns = 24 hours)
heatmap = np.zeros((len(participants), 24), dtype=int)

# Fill the heatmap
for msg in messages:
    sender = msg["sender"]
    timestamp = msg["timestamp"]

    date, time = timestamp.split(", ")
    hour = int(time.split(":")[0])

    row = person_index[sender]
    heatmap[row][hour] += 1

print("Heatmap Matrix:")
print(heatmap)
```

```python

print("=" * 60)
print("ACTIVITY HEATMAP")
print("=" * 60)

print("Hour ->", end=" ")
for h in range(24):
    print(f"{h:02}", end=" ")
print()

for i, name in enumerate(participants):
    print(f"{name:<10}", end=" ")

    for value in heatmap[i]:
        if value == 0:
            symbol = "."
        elif value <= 5:
            symbol = "░"
        elif value <= 10:
            symbol = "▒"
        else:
            symbol = "█"

        print(symbol, end="  ")

    print()
```

```python

stop_words = {
    "the","is","a","an","and","or","to","of","in","on","for",
    "i","you","me","my","we","our","are","it","this","that"
}

word_count = {}

for msg in messages:
    text = msg["message"].lower()

    for ch in ".,!?;:()[]{}\"'":
        text = text.replace(ch, "")

    words = text.split()

    for word in words:
        if word not in stop_words and word != "<media":
            word_count[word] = word_count.get(word, 0) + 1

top10 = sorted(word_count.items(), key=lambda x: x[1], reverse=True)[:10]

print("="*50)
print("TOP 10 WORDS")
print("="*50)

for word, count in top10:
    print(f"{word:<15} {'█'*min(count//5,20)} {count}")
```

```python

from datetime import datetime

response = {}

for i in range(1, len(messages)):
    prev = messages[i-1]
    curr = messages[i]

    if prev["sender"] != curr["sender"]:
        try:
            t1 = datetime.strptime(prev["timestamp"], "%d/%m/%y, %H:%M")
            t2 = datetime.strptime(curr["timestamp"], "%d/%m/%y, %H:%M")

            diff = (t2 - t1).total_seconds() / 60

            response.setdefault(curr["sender"], []).append(diff)
        except ValueError:
            continue

print("Average Response Time")

for person, times in response.items():
    if times:
        avg = sum(times) / len(times)
        print(f"{person}: {avg:.2f} minutes")
```

```python

activity = {}

for msg in messages:

    person = msg["sender"]

    date = msg["timestamp"].split(",")[0]

    activity.setdefault(person, set()).add(date)

print("="*50)
print("ACTIVE DAYS")
print("="*50)

for person in activity:

    print(f"{person:<12}{len(activity[person])} active days")
```

```python

print("="*60)
print("PERSONALITY ARCHETYPES")
print("="*60)

for i, person in enumerate(participants):

    total = sum(heatmap[i])

    # Night Owl
    night = sum(heatmap[i][23:24]) + sum(heatmap[i][0:5])

    if total > 0:
        night_percent = (night / total) * 100
    else:
        night_percent = 0

    # Average words per message
    person_msgs = [m for m in messages if m["sender"] == person]

    avg_words = 0
    if person_msgs:
        avg_words = sum(len(m["message"].split()) for m in person_msgs) / len(person_msgs)

    # Active days
    active_days = len(activity.get(person, set()))

    # Decide archetype
    if night_percent >= 60:
        archetype = "🌙 THE NIGHT OWL"
    elif avg_words >= 30:
        archetype = "📖 THE STORYTELLER"
    elif active_days <= 20:
        archetype = "👻 THE GHOST"
    elif message_count[person] == max(message_count.values()):
        archetype = "💬 THE SPAMMER"
    else:
        archetype = "🙂 THE NORMAL MEMBER"

    print(f"{person:<15} {archetype}")
```
